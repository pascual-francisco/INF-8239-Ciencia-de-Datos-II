# DECISION RECORD

## Resumen

Se realizó una evaluación de impacto ético para el sistema de apoyo a decisiones propuesto. Los riesgos fueron evaluados utilizando probabilidad e impacto, y se identificaron los riesgos con mayor prioridad para determinar si el sistema puede continuar su desarrollo y uso.

## Riesgos prioritarios

### R02 - Clasificación incorrecta de territorios prioritarios

Este es el riesgo con mayor prioridad. Una clasificación errónea puede provocar que recursos limitados sean asignados a territorios con menor necesidad, mientras que áreas con necesidades críticas podrían quedar sin atención. La revisión humana de los resultados ayuda a reducir este riesgo, pero requiere monitoreo continuo de la calidad de los datos y del desempeño del sistema.

### R01 - Sesgo sistemático en la clasificación de territorios

Existe la posibilidad de que ciertos territorios sean favorecidos o perjudicados debido a patrones presentes en los datos utilizados para la evaluación. Este riesgo puede afectar la equidad de las decisiones tomadas. Como medida de control se establecieron auditorías periódicas de equidad y revisión de métricas por grupos afectados.

### R07 - Dependencia excesiva de recomendaciones automatizadas

Los operadores podrían confiar excesivamente en las recomendaciones generadas por el sistema y disminuir el juicio profesional en la toma de decisiones. Para mitigar este riesgo se requiere mantener una revisión humana obligatoria antes de ejecutar cualquier acción basada en las recomendaciones del sistema.

## Incertidumbre

Aunque existen controles documentados, permanece incertidumbre respecto a cambios futuros en los datos, cambios regulatorios y posibles desviaciones del comportamiento esperado del sistema. Por esta razón se recomienda un monitoreo periódico y revisiones continuas de desempeño.

## Responsable

El propietario del sistema es el Departamento de Planificación Territorial, con apoyo del equipo de datos para monitoreo, auditoría y mantenimiento de controles.

## Recomendación

### GO / CONTINUAR

Se recomienda **continuar** con el sistema debido a que los riesgos identificados cuentan con controles definidos, responsables asignados y mecanismos de supervisión. No obstante, el uso del sistema debe mantenerse sujeto a revisión humana, auditorías periódicas y cumplimiento de las condiciones de retiro establecidas en la evaluación de impacto ético.

```python
# ==============================================================================
# AUTOMATIC COMMIT AND AUTHENTICATED PUSH
# ==============================================================================

import os
import stat
import subprocess
from pathlib import Path
from google.colab import userdata

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

REPO_PATH = Path(
    "/content/INF-8239-Ciencia-de-Datos-II"
)

GITHUB_USER = "pascual-francisco"
GITHUB_REPO = "INF-8239-Ciencia-de-Datos-II"

COMMIT_MESSAGE = "chore: sync repository updates"

GITHUB_TOKEN = userdata.get("TOKEN")

# ------------------------------------------------------------------------------

def run_git(
    arguments,
    *,
    check=True,
    capture_output=False,
    environment=None
):
    return subprocess.run(
        ["git", *arguments],
        cwd=REPO_PATH,
        check=check,
        text=True,
        capture_output=capture_output,
        env=environment
    )

# ------------------------------------------------------------------------------
# VALIDATE TOKEN
# ------------------------------------------------------------------------------

if not GITHUB_TOKEN:
    raise ValueError(
        "TOKEN not found in Colab Secrets."
    )

# ------------------------------------------------------------------------------
# STAGE CHANGES
# ------------------------------------------------------------------------------

print("=" * 70)
print("STAGING CHANGES")
print("=" * 70)

run_git(["add", "-A"])

# ------------------------------------------------------------------------------
# CREATE COMMIT IF NEEDED
# ------------------------------------------------------------------------------

staged = run_git(
    ["diff", "--cached", "--name-status"],
    capture_output=True
)

staged_output = staged.stdout.strip()

if staged_output:

    print(staged_output)

    run_git(
        [
            "commit",
            "-m",
            COMMIT_MESSAGE
        ]
    )

    print(
        f"\nCommit created: {COMMIT_MESSAGE}"
    )

else:

    print(
        "\nNo new changes to commit."
    )

# ------------------------------------------------------------------------------
# CURRENT BRANCH
# ------------------------------------------------------------------------------

branch = run_git(
    [
        "branch",
        "--show-current"
    ],
    capture_output=True
).stdout.strip()

print(f"\nCurrent branch: {branch}")

# ------------------------------------------------------------------------------
# CLEAN REMOTE
# ------------------------------------------------------------------------------

CLEAN_REMOTE = (
    f"https://github.com/{GITHUB_USER}/{GITHUB_REPO}.git"
)

run_git(
    [
        "remote",
        "set-url",
        "origin",
        CLEAN_REMOTE
    ]
)

# ------------------------------------------------------------------------------
# TEMPORARY AUTHENTICATION
# ------------------------------------------------------------------------------

askpass_path = Path(
    "/tmp/github_colab_askpass.sh"
)

askpass_path.write_text(
    """#!/bin/sh
case "$1" in
    *Username*) printf '%s\\n' "$GITHUB_USERNAME" ;;
    *Password*) printf '%s\\n' "$GITHUB_TOKEN" ;;
    *) printf '\\n' ;;
esac
""",
    encoding="utf-8"
)

askpass_path.chmod(
    stat.S_IRUSR
    | stat.S_IWUSR
    | stat.S_IXUSR
)

push_env = os.environ.copy()

push_env["GIT_ASKPASS"] = str(
    askpass_path
)

push_env["GIT_TERMINAL_PROMPT"] = "0"

push_env["GITHUB_USERNAME"] = (
    GITHUB_USER
)

push_env["GITHUB_TOKEN"] = (
    GITHUB_TOKEN
)

# ------------------------------------------------------------------------------
# PUSH
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("PUSHING TO GITHUB")
print("=" * 70)

try:

    run_git(
        [
            "push",
            "--set-upstream",
            "origin",
            branch
        ],
        environment=push_env
    )

    print("\nPush completed successfully.")

finally:

    if askpass_path.exists():
        askpass_path.unlink()

# ------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL STATUS")
print("=" * 70)

run_git(
    [
        "status",
        "--short",
        "--branch"
    ]
)

print("\nRepository synchronized successfully.")
```
