---

## Accepted Dataset

### Spanish Tweet Datasets Encoded for Sentiment Analysis Based on Context of Words

The dataset published by the Universidad de Alcalá was accepted because it contains Spanish-language tweets with explicit positive, neutral, and negative sentiment labels.

The corpus has documented academic provenance, an identifiable version, a persistent DOI, reproducible public access, and a Creative Commons Attribution 4.0 International (`CC BY 4.0`) license.

The dataset includes the following original labels:

- `P`: Positive sentiment
- `NEU`: Neutral sentiment
- `N`: Negative sentiment
- `NONE`: No expressed opinion

For the proposed three-class sentiment-analysis task, the processed dataset will retain `P`, `NEU`, and `N`. The `NONE` category will be excluded because absence of opinion is conceptually different from neutral sentiment.

The principal limitation is the severe underrepresentation of the neutral class. In the General Corpus, neutral observations represent approximately 2.25% of the data. Therefore, accuracy alone may provide a misleading evaluation. Macro F1-score, recall by class, and the confusion matrix should be prioritized.

**Decision: Accepted with conditions.**

---

## Rejected Dataset

### TwitterSentimentDataset

TwitterSentimentDataset was considered because it contains a large number of naturally occurring Spanish-language tweets and is publicly available under the Apache License 2.0.

The corpus contains approximately 250,000 tweets divided into positive, negative, and unclassified-sentiment groups. Positive and negative labels were generated automatically using emoticons as sentiment indicators.

The dataset was rejected for the following reasons:

1. It does not provide an explicit neutral sentiment class.
2. Tweets without a positive or negative emoticon are not necessarily neutral.
3. A tweet can express sentiment without using an emoticon.
4. Emoticons can be used ironically or sarcastically.
5. The automatic labeling procedure may introduce substantial label noise.
6. The corpus may overrepresent users and messages that use emoticons.
7. An institutional DOI and formally documented dataset version were not identified.
8. The proposed research question requires an interpretable neutral class.

Although the dataset is larger, dataset size does not compensate for weak target-label validity. The accepted corpus provides labels that better correspond to the intended three-class classification task.

**Decision: Rejected because it does not provide a reliable neutral class.**

---

````python
# ==============================================================================
# CELL 17 — RECONSTRUCT AND AUDIT THE DOWNLOADED PUBLIC CORPUS
# Objective: Reconstruct the original texts, preserve all documented records,
# distinguish unlabeled evaluation data, and audit the downloaded corpus.
# ==============================================================================

# Import hashlib to calculate SHA-256 file identifiers.
import hashlib

# Import Path for operating-system-independent path management.
from pathlib import Path

# Import pandas for creating and auditing the reconstructed dataset.
import pandas as pd


# ------------------------------------------------------------------------------
# 1. DEFINE THE CORPUS FILES
# ------------------------------------------------------------------------------

# Define the project root directory.
PROJECT_PATH = Path(
    "/content/INF-8239-Ciencia-de-Datos-II/"
    "Unit_02_Natural_Language_Processing/"
    "nlp_sentiment_analysis"
)

# Define the General Corpus directory.
CORPUS_PATH = PROJECT_PATH / "data" / "raw" / "Data" / "Corpus_gral"

# Define the original downloaded archive.
ZIP_PATH = PROJECT_PATH / "data" / "raw" / "Data.zip"

# Define the files required to reconstruct the corpus.
METADATA_PATH = CORPUS_PATH / "reviews_tw_definitivo.txt"
ENCODED_REVIEWS_PATH = CORPUS_PATH / "reviews_cod.txt"
DICTIONARY_PATH = CORPUS_PATH / "words_cod.txt"

# Verify that every required file exists.
required_files = [
    ZIP_PATH,
    METADATA_PATH,
    ENCODED_REVIEWS_PATH,
    DICTIONARY_PATH
]

for file_path in required_files:
    if not file_path.exists():
        raise FileNotFoundError(
            f"Required corpus file not found:\n{file_path}"
        )


# ------------------------------------------------------------------------------
# 2. DEFINE THE SHA-256 FUNCTION
# ------------------------------------------------------------------------------

def calculate_sha256(file_path: Path) -> str:
    """Calculate the SHA-256 hash of a file."""

    # Create a SHA-256 hashing object.
    sha256_hash = hashlib.sha256()

    # Read the file in binary chunks.
    with file_path.open("rb") as file_object:
        for chunk in iter(lambda: file_object.read(8192), b""):
            sha256_hash.update(chunk)

    # Return the final hexadecimal identifier.
    return sha256_hash.hexdigest()


# ------------------------------------------------------------------------------
# 3. LOAD THE WORD DICTIONARY
# ------------------------------------------------------------------------------

# Create a mapping from word identifiers to original words.
id_to_word = {}

# Read the word dictionary.
with DICTIONARY_PATH.open("r", encoding="utf-8") as dictionary_file:

    for line in dictionary_file:
        # Remove surrounding whitespace.
        line = line.strip()

        # Ignore empty or malformed lines.
        if not line or ";" not in line:
            continue

        # Separate the word from its identifier.
        word, word_identifier = line.rsplit(";", 1)

        # Store the dictionary relationship.
        id_to_word[int(word_identifier)] = word

print(f"Dictionary entries loaded: {len(id_to_word):,}")


# ------------------------------------------------------------------------------
# 4. LOAD THE REVIEW METADATA
# ------------------------------------------------------------------------------

# Define the dataset-set mapping identified from the documentation and files.
set_mapping = {
    1: "training",
    2: "test",
    3: "evaluation",
    5: "general"
}

# Define the sentiment-label mapping.
# Code 5 is not a sentiment class; it identifies unavailable evaluation labels.
label_mapping = {
    1: "negative",
    2: "neutral",
    3: "positive",
    4: "none",
    5: "unlabeled"
}

# Load the metadata file.
metadata_df = pd.read_csv(
    METADATA_PATH,
    sep=";",
    header=None,
    names=[
        "identifier",
        "set_code",
        "label_code"
    ]
)

# Convert the documented codes into readable categories.
metadata_df["dataset_set"] = metadata_df["set_code"].map(set_mapping)
metadata_df["label"] = metadata_df["label_code"].map(label_mapping)

# Verify that every code was mapped successfully.
if metadata_df["dataset_set"].isna().any():
    raise ValueError(
        "One or more dataset-set codes are not documented."
    )

if metadata_df["label"].isna().any():
    raise ValueError(
        "One or more label codes are not documented."
    )

print(f"Metadata records loaded: {len(metadata_df):,}")


# ------------------------------------------------------------------------------
# 5. RECONSTRUCT THE TEXTS
# ------------------------------------------------------------------------------

# Create a list for the reconstructed texts.
reconstructed_texts = []

# Store the words of the current review.
current_review_words = []

# Count word identifiers not found in the dictionary.
unknown_word_identifiers = 0

# Read the encoded corpus.
with ENCODED_REVIEWS_PATH.open(
    "r",
    encoding="utf-8"
) as encoded_reviews_file:

    for line in encoded_reviews_file:
        # Remove surrounding whitespace.
        line = line.strip()

        # The %% line marks the end of a review.
        if line == "%%":
            reconstructed_texts.append(
                " ".join(current_review_words).strip()
            )

            current_review_words = []
            continue

        # Ignore empty lines.
        if not line:
            continue

        # The first field contains the current word identifier.
        current_word_identifier = int(
            line.split(";", 1)[0]
        )

        # Retrieve the original word.
        current_word = id_to_word.get(
            current_word_identifier
        )

        # Add known words to the current review.
        if current_word is not None:
            current_review_words.append(current_word)
        else:
            unknown_word_identifiers += 1

# Preserve the final review if the file does not end with %%.
if current_review_words:
    reconstructed_texts.append(
        " ".join(current_review_words).strip()
    )

print(f"Reviews reconstructed: {len(reconstructed_texts):,}")
print(f"Unknown word identifiers: {unknown_word_identifiers:,}")


# ------------------------------------------------------------------------------
# 6. VERIFY THE TEXT-METADATA RELATIONSHIP
# ------------------------------------------------------------------------------

# Confirm that every metadata record has one reconstructed text.
if len(reconstructed_texts) != len(metadata_df):
    raise ValueError(
        "The number of reconstructed reviews does not match the metadata.\n"
        f"Reconstructed reviews: {len(reconstructed_texts):,}\n"
        f"Metadata records:      {len(metadata_df):,}"
    )

# Add the reconstructed texts to the metadata.
corpus_df = metadata_df.copy()
corpus_df["text"] = reconstructed_texts

# Organize the dataset columns.
corpus_df = corpus_df[
    [
        "identifier",
        "set_code",
        "dataset_set",
        "text",
        "label_code",
        "label"
    ]
]


# ------------------------------------------------------------------------------
# 7. REPORT FILE IDENTITY AND STRUCTURE
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("FILE AND DATASET STRUCTURE")
print("=" * 70)
print(f"Original archive: {ZIP_PATH}")
print(f"Archive SHA-256:  {calculate_sha256(ZIP_PATH)}")
print(f"Rows:             {corpus_df.shape,}")
print(f"Columns:          {corpus_df.shape,}")

print("\nColumn names:")

for column_name in corpus_df.columns:
    print(f"- {column_name}")


# ------------------------------------------------------------------------------
# 8. REPORT MISSING AND EMPTY VALUES
# ------------------------------------------------------------------------------

# Calculate missing values.
missing_values = corpus_df.isna().sum()

# Calculate empty reconstructed texts.
empty_text_count = (
    corpus_df["text"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("\n" + "=" * 70)
print("MISSING AND EMPTY VALUES")
print("=" * 70)
print(missing_values.to_string())
print(f"\nEmpty or whitespace-only texts: {empty_text_count:,}")


# ------------------------------------------------------------------------------
# 9. REPORT DUPLICATES
# ------------------------------------------------------------------------------

# Calculate complete duplicate rows.
duplicate_rows = corpus_df.duplicated().sum()

# Calculate additional duplicated texts.
duplicate_texts = corpus_df.duplicated(
    subset=["text"]
).sum()

# Calculate additional duplicated text-label pairs.
duplicate_text_label_pairs = corpus_df.duplicated(
    subset=["text", "label"]
).sum()

print("\n" + "=" * 70)
print("DUPLICATES")
print("=" * 70)
print(f"Duplicate complete rows:       {duplicate_rows:,}")
print(f"Duplicate text occurrences:    {duplicate_texts:,}")
print(f"Duplicate text-label pairs:    {duplicate_text_label_pairs:,}")


# ------------------------------------------------------------------------------
# 10. REPORT PARTITION DISTRIBUTION
# ------------------------------------------------------------------------------

# Calculate the number of observations in each dataset partition.
set_distribution = pd.crosstab(
    corpus_df["dataset_set"],
    corpus_df["label"],
    margins=True
)

print("\n" + "=" * 70)
print("DATASET SET AND LABEL DISTRIBUTION")
print("=" * 70)
display(set_distribution)


# ------------------------------------------------------------------------------
# 11. REPORT THE COMPLETE LABEL DISTRIBUTION
# ------------------------------------------------------------------------------

# Calculate complete counts and percentages, including unlabeled observations.
all_label_counts = corpus_df["label"].value_counts(
    dropna=False
)

all_label_percentages = (
    corpus_df["label"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
)

all_label_distribution = pd.DataFrame({
    "count": all_label_counts,
    "percentage": all_label_percentages
})

print("\n" + "=" * 70)
print("COMPLETE LABEL DISTRIBUTION")
print("=" * 70)
display(all_label_distribution.round(2))


# ------------------------------------------------------------------------------
# 12. REPORT THE SUPERVISED LABEL DISTRIBUTION
# ------------------------------------------------------------------------------

# Exclude records whose sentiment label is unavailable.
labeled_corpus_df = corpus_df[
    corpus_df["label"] != "unlabeled"
].copy()

# Calculate the distribution of documented sentiment labels.
supervised_label_counts = (
    labeled_corpus_df["label"]
    .value_counts()
)

supervised_label_percentages = (
    labeled_corpus_df["label"]
    .value_counts(normalize=True)
    .mul(100)
)

supervised_label_distribution = pd.DataFrame({
    "count": supervised_label_counts,
    "percentage": supervised_label_percentages
})

print("\n" + "=" * 70)
print("SUPERVISED LABEL DISTRIBUTION")
print("=" * 70)
display(supervised_label_distribution.round(2))


# ------------------------------------------------------------------------------
# 13. REPORT THE THREE-CLASS DISTRIBUTION
# ------------------------------------------------------------------------------

# Keep only the target classes proposed for the future model.
three_class_df = corpus_df[
    corpus_df["label"].isin(
        ["negative", "neutral", "positive"]
    )
].copy()

# Calculate the three-class distribution.
three_class_counts = three_class_df["label"].value_counts()

three_class_percentages = (
    three_class_df["label"]
    .value_counts(normalize=True)
    .mul(100)
)

three_class_distribution = pd.DataFrame({
    "count": three_class_counts,
    "percentage": three_class_percentages
})

print("\n" + "=" * 70)
print("PROPOSED THREE-CLASS DISTRIBUTION")
print("=" * 70)
display(three_class_distribution.round(2))


# ------------------------------------------------------------------------------
# 14. REPORT TEXT-LENGTH STATISTICS
# ------------------------------------------------------------------------------

# Calculate text lengths in characters.
text_lengths = corpus_df["text"].str.len()

# Calculate the requested percentiles.
text_length_summary = text_lengths.describe(
    percentiles=[0.50, 0.90, 0.99]
)

print("\n" + "=" * 70)
print("TEXT LENGTHS IN CHARACTERS")
print("=" * 70)
print(text_length_summary.to_string())


# ------------------------------------------------------------------------------
# 15. DISPLAY RECONSTRUCTED EXAMPLES
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("RECONSTRUCTED DATA SAMPLE")
print("=" * 70)

# Display examples from the reconstructed corpus.
display(
    corpus_df[
        [
            "identifier",
            "dataset_set",
            "text",
            "label"
        ]
    ].head()
)


# ------------------------------------------------------------------------------
# 16. COMPLETE THE AUDIT
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("DOWNLOADED PUBLIC CORPUS AUDIT COMPLETED")
print("=" * 70)
````

## Downloaded Corpus Audit Results

The downloaded corpus was reconstructed from its encoded representation using the provided word dictionary, review separators, metadata, and sentiment-label codes.

A total of **63,238 records** were reconstructed, matching the number of metadata observations. No missing values were detected in the reconstructed columns. However, **114 records contained empty or whitespace-only text** and should be excluded from the processed modeling dataset.

The audit identified **1,599 duplicated text occurrences** and **1,565 duplicated text-label pairs**. These duplicates must be controlled before creating or modifying model partitions because duplicated observations across training and test sets could artificially inflate model performance.

The complete corpus contains 22,356 positive records, 16,219 negative records, 1,486 neutral records, 21,278 records labeled as no opinion, and 1,899 evaluation records without disclosed sentiment labels.

After preserving only the proposed negative, neutral, and positive classes, the dataset contains **40,061 labeled observations**. Positive represents 55.80%, negative represents 40.49%, and neutral represents only 3.71%. The severe underrepresentation of neutral sentiment is the principal class-quality limitation.

The reconstruction also encountered **203,114 unknown word-identifier occurrences** that could not be recovered using the provided dictionary. As a result, the reconstructed text may omit some original tokens. This limitation must be documented in the dataset card.

Text lengths range from 0 to 315 characters. The median length is 89 characters, while the 90th and 99th percentiles are 124 and 139 characters, respectively.

The corpus is approved with conditions. Empty texts, records without disclosed labels, the no-opinion category, and duplicate texts must be treated explicitly before modeling.

## Text Reconstruction

The downloaded corpus stores each review as a sequence of encoded word
identifiers. The original text was reconstructed by mapping each `id_Dict`
value to the corresponding word in the supplied dictionary and joining the
tokens until the `%%` review separator was reached.

The audit detected 203,114 occurrences of identifiers not present in the
dictionary. Further analysis showed that every occurrence corresponded to
the same identifier: `0`.

Identifier `0` was interpreted as a reserved or padding value rather than a
recoverable lexical token. It was therefore excluded during text
reconstruction. All positive word identifiers were successfully resolved
using the supplied dictionary.

This means that the audit did not identify 203,114 different unknown words.
It identified one reserved code repeated 203,114 times.

````python
# ==============================================================================
# LAB05 — CELL 05: INTERPRET MODEL METRICS
# Objective: Load the generated evaluation results, compare all classifiers,
# and answer the required model-interpretation questions automatically.
# ==============================================================================

# Import Path for operating-system-independent path management.
from pathlib import Path

# Import pandas for loading, organizing, and displaying evaluation results.
import pandas as pd


# ------------------------------------------------------------------------------
# 1. DEFINE THE PROJECT REPORT PATHS
# ------------------------------------------------------------------------------

# Define the root directory of the NLP sentiment-analysis project.
PROJECT_PATH = Path(
    "/content/INF-8239-Ciencia-de-Datos-II/"
    "Unit_02_Natural_Language_Processing/"
    "nlp_sentiment_analysis"
)

# Define the metrics file generated by scripts/train_text.py.
METRICS_PATH = (
    PROJECT_PATH
    / "reports"
    / "text_metrics.csv"
)

# Define the error-analysis file generated by scripts/train_text.py.
ERROR_ANALYSIS_PATH = (
    PROJECT_PATH
    / "reports"
    / "error_analysis.csv"
)

# Define the confusion-matrix image generated by scripts/train_text.py.
CONFUSION_MATRIX_PATH = (
    PROJECT_PATH
    / "reports"
    / "confusion_text.png"
)


# ------------------------------------------------------------------------------
# 2. VERIFY AND LOAD THE MODEL METRICS
# ------------------------------------------------------------------------------

# Stop execution when the metrics file is unavailable.
if not METRICS_PATH.exists():
    raise FileNotFoundError(
        "The model metrics file was not found:\n"
        f"{METRICS_PATH}\n\n"
        "Run scripts/train_text.py before executing this cell."
    )

# Load the model evaluation results.
metrics_df = pd.read_csv(METRICS_PATH)

# Remove surrounding whitespace from column names.
metrics_df.columns = (
    metrics_df.columns
    .astype(str)
    .str.strip()
)


# ------------------------------------------------------------------------------
# 3. DEFINE A COLUMN-DETECTION FUNCTION
# ------------------------------------------------------------------------------

def find_column(
    dataframe: pd.DataFrame,
    possible_names: list[str]
) -> str | None:
    """
    Find the first DataFrame column matching one of the possible names.

    Parameters
    ----------
    dataframe:
        DataFrame whose columns will be inspected.
    possible_names:
        Alternative names that may represent the required variable.

    Returns
    -------
    str or None
        Matching column name, or None when no match is found.
    """

    # Create a lowercase mapping while preserving the original column names.
    lowercase_columns = {
        column_name.lower().strip(): column_name
        for column_name in dataframe.columns
    }

    # Check every possible column name.
    for possible_name in possible_names:
        normalized_name = possible_name.lower().strip()

        if normalized_name in lowercase_columns:
            return lowercase_columns[normalized_name]

    # Return None when no compatible column is found.
    return None


# ------------------------------------------------------------------------------
# 4. IDENTIFY THE AVAILABLE METRIC COLUMNS
# ------------------------------------------------------------------------------

# Identify the column containing the model name.
model_column = find_column(
    metrics_df,
    [
        "model",
        "model_name",
        "classifier",
        "pipeline"
    ]
)

# Identify the column containing macro F1-score.
macro_f1_column = find_column(
    metrics_df,
    [
        "macro_f1",
        "f1_macro",
        "macro f1",
        "macro_f1_score"
    ]
)

# Identify an optional class or label column.
class_column = find_column(
    metrics_df,
    [
        "class",
        "label",
        "target",
        "sentiment"
    ]
)

# Identify optional per-class evaluation columns.
precision_column = find_column(
    metrics_df,
    [
        "precision",
        "class_precision"
    ]
)

recall_column = find_column(
    metrics_df,
    [
        "recall",
        "class_recall"
    ]
)

f1_column = find_column(
    metrics_df,
    [
        "f1",
        "f1_score",
        "class_f1"
    ]
)

support_column = find_column(
    metrics_df,
    [
        "support",
        "class_support"
    ]
)

# Stop execution if the essential comparison columns are unavailable.
if model_column is None:
    raise KeyError(
        "A model-name column was not found.\n"
        f"Available columns: {metrics_df.columns.tolist()}"
    )

if macro_f1_column is None:
    raise KeyError(
        "A macro F1 column was not found.\n"
        f"Available columns: {metrics_df.columns.tolist()}"
    )


# ------------------------------------------------------------------------------
# 5. DISPLAY THE METRICS FILE INFORMATION
# ------------------------------------------------------------------------------

print("=" * 70)
print("MODEL METRICS FILE")
print("=" * 70)
print(f"Path: {METRICS_PATH}")
print(f"Rows: {metrics_df.shape,}")
print(f"Columns: {metrics_df.shape,}")

print("\nAvailable metric columns:")

for column_name in metrics_df.columns:
    print(f"- {column_name}")


# ------------------------------------------------------------------------------
# 6. PREPARE THE MODEL-LEVEL COMPARISON
# ------------------------------------------------------------------------------

# Create one model-level row per classifier.
model_comparison_df = (
    metrics_df[
        [
            model_column,
            macro_f1_column
        ]
    ]
    .dropna()
    .drop_duplicates(
        subset=[model_column]
    )
    .copy()
)

# Convert macro F1-score into a numeric variable.
model_comparison_df[macro_f1_column] = pd.to_numeric(
    model_comparison_df[macro_f1_column],
    errors="coerce"
)

# Remove rows whose macro F1-score could not be interpreted numerically.
model_comparison_df = model_comparison_df.dropna(
    subset=[macro_f1_column]
)

# Sort classifiers from highest to lowest macro F1-score.
model_comparison_df = model_comparison_df.sort_values(
    by=macro_f1_column,
    ascending=False
).reset_index(drop=True)

# Stop when no valid results remain.
if model_comparison_df.empty:
    raise ValueError(
        "No valid model-level macro F1 results were found."
    )

print("\n" + "=" * 70)
print("MODEL COMPARISON BY MACRO F1")
print("=" * 70)

display(model_comparison_df)


# ------------------------------------------------------------------------------
# 7. IDENTIFY THE BEST MODEL AND THE BASELINE
# ------------------------------------------------------------------------------

# Select the classifier with the highest macro F1-score.
best_model_name = str(
    model_comparison_df.iloc[0][model_column]
)

best_macro_f1 = float(
    model_comparison_df.iloc[0][macro_f1_column]
)

# Detect the baseline using common DummyClassifier or baseline names.
baseline_mask = (
    model_comparison_df[model_column]
    .astype(str)
    .str.lower()
    .str.contains(
        "dummy|baseline",
        regex=True
    )
)

# Store baseline information when available.
if baseline_mask.any():
    baseline_row = model_comparison_df[
        baseline_mask
    ].iloc[0]

    baseline_model_name = str(
        baseline_row[model_column]
    )

    baseline_macro_f1 = float(
        baseline_row[macro_f1_column]
    )

    macro_f1_improvement = (
        best_macro_f1 - baseline_macro_f1
    )

else:
    baseline_model_name = None
    baseline_macro_f1 = None
    macro_f1_improvement = None


# ------------------------------------------------------------------------------
# 8. IDENTIFY THE CLASS WITH THE LOWEST RECALL
# ------------------------------------------------------------------------------

lowest_recall_class = None
lowest_recall_value = None
lowest_recall_support = None

# Continue only when class-level recall is available.
if class_column is not None and recall_column is not None:

    # Select the rows belonging to the best model.
    best_model_class_metrics = metrics_df[
        metrics_df[model_column].astype(str).eq(
            best_model_name
        )
    ].copy()

    # Keep only the three target sentiment classes.
    sentiment_classes = [
        "negative",
        "neutral",
        "positive"
    ]

    best_model_class_metrics = best_model_class_metrics[
        best_model_class_metrics[class_column]
        .astype(str)
        .str.lower()
        .isin(sentiment_classes)
    ].copy()

    # Convert recall into a numeric column.
    best_model_class_metrics[recall_column] = pd.to_numeric(
        best_model_class_metrics[recall_column],
        errors="coerce"
    )

    # Remove rows without a valid recall value.
    best_model_class_metrics = best_model_class_metrics.dropna(
        subset=[recall_column]
    )

    # Identify the class with the lowest recall.
    if not best_model_class_metrics.empty:
        lowest_recall_row = best_model_class_metrics.sort_values(
            by=recall_column,
            ascending=True
        ).iloc[0]

        lowest_recall_class = str(
            lowest_recall_row[class_column]
        )

        lowest_recall_value = float(
            lowest_recall_row[recall_column]
        )

        # Retrieve support when it is available.
        if support_column is not None:
            lowest_recall_support = float(
                lowest_recall_row[support_column]
            )


# ------------------------------------------------------------------------------
# 9. IDENTIFY THE CLASS WITH THE LOWEST PRECISION
# ------------------------------------------------------------------------------

lowest_precision_class = None
lowest_precision_value = None

# Continue only when class-level precision is available.
if class_column is not None and precision_column is not None:

    # Select class-level metrics for the best classifier.
    best_model_precision_metrics = metrics_df[
        metrics_df[model_column].astype(str).eq(
            best_model_name
        )
    ].copy()

    # Keep only the three sentiment classes.
    best_model_precision_metrics = (
        best_model_precision_metrics[
            best_model_precision_metrics[class_column]
            .astype(str)
            .str.lower()
            .isin(
                [
                    "negative",
                    "neutral",
                    "positive"
                ]
            )
        ]
        .copy()
    )

    # Convert precision into a numeric variable.
    best_model_precision_metrics[precision_column] = pd.to_numeric(
        best_model_precision_metrics[precision_column],
        errors="coerce"
    )

    # Remove invalid values.
    best_model_precision_metrics = (
        best_model_precision_metrics.dropna(
            subset=[precision_column]
        )
    )

    # Identify the class with the lowest precision.
    if not best_model_precision_metrics.empty:
        lowest_precision_row = (
            best_model_precision_metrics.sort_values(
                by=precision_column,
                ascending=True
            )
            .iloc[0]
        )

        lowest_precision_class = str(
            lowest_precision_row[class_column]
        )

        lowest_precision_value = float(
            lowest_precision_row[precision_column]
        )


# ------------------------------------------------------------------------------
# 10. INSPECT OPTIONAL ERROR-ANALYSIS RESULTS
# ------------------------------------------------------------------------------

error_analysis_df = None
true_label_column = None
predicted_label_column = None
largest_confusion_description = None
most_frequent_error_type = None

# Load the error-analysis file when available.
if ERROR_ANALYSIS_PATH.exists():

    error_analysis_df = pd.read_csv(
        ERROR_ANALYSIS_PATH
    )

    error_analysis_df.columns = (
        error_analysis_df.columns
        .astype(str)
        .str.strip()
    )

    # Identify the real and predicted label columns.
    true_label_column = find_column(
        error_analysis_df,
        [
            "true_label",
            "actual_label",
            "y_true",
            "true",
            "actual"
        ]
    )

    predicted_label_column = find_column(
        error_analysis_df,
        [
            "predicted_label",
            "prediction",
            "y_pred",
            "predicted"
        ]
    )

    # Identify the most frequent real-to-predicted confusion.
    if (
        true_label_column is not None
        and predicted_label_column is not None
        and not error_analysis_df.empty
    ):
        confusion_counts = (
            error_analysis_df
            .groupby(
                [
                    true_label_column,
                    predicted_label_column
                ]
            )
            .size()
            .sort_values(
                ascending=False
            )
        )

        if not confusion_counts.empty:
            (
                most_confused_true_class,
                most_confused_predicted_class
            ) = confusion_counts.index[0]

            largest_confusion_count = int(
                confusion_counts.iloc[0]
            )

            largest_confusion_description = (
                f"{largest_confusion_count:,} errors changed "
                f"'{most_confused_true_class}' into "
                f"'{most_confused_predicted_class}'."
            )

    # Identify an optional manually assigned error-category column.
    error_category_column = find_column(
        error_analysis_df,
        [
            "error_category",
            "category",
            "error_type"
        ]
    )

    if error_category_column is not None:
        categorized_errors = (
            error_analysis_df[error_category_column]
            .dropna()
            .astype(str)
            .str.strip()
        )

        categorized_errors = categorized_errors[
            categorized_errors.ne("")
        ]

        if not categorized_errors.empty:
            most_frequent_error_type = str(
                categorized_errors.value_counts().index[0]
            )


# ------------------------------------------------------------------------------
# 11. ANSWER THE REQUIRED INTERPRETATION QUESTIONS
# ------------------------------------------------------------------------------

questions_and_answers = []

# Question 1: Does the selected classifier exceed the baseline?
if baseline_macro_f1 is not None:
    if macro_f1_improvement > 0:
        baseline_answer = (
            f"Yes. {best_model_name} obtained a macro F1 of "
            f"{best_macro_f1:.4f}, compared with "
            f"{baseline_macro_f1:.4f} for {baseline_model_name}. "
            f"The absolute improvement was "
            f"{macro_f1_improvement:.4f}."
        )
    else:
        baseline_answer = (
            f"No meaningful improvement was observed. "
            f"{best_model_name} obtained {best_macro_f1:.4f}, "
            f"while {baseline_model_name} obtained "
            f"{baseline_macro_f1:.4f}."
        )
else:
    baseline_answer = (
        "A baseline row could not be identified automatically. "
        "Verify that the DummyClassifier model name contains "
        "'dummy' or 'baseline' in text_metrics.csv."
    )

questions_and_answers.append({
    "Question": "Does the selected model exceed the baseline?",
    "Answer": baseline_answer
})


# Question 2: Which class has the greatest recall difficulty?
if lowest_recall_class is not None:
    recall_answer = (
        f"The class with the lowest recall was "
        f"'{lowest_recall_class}', with a recall of "
        f"{lowest_recall_value:.4f}. This means that the model "
        f"failed to recover a larger proportion of the real "
        f"{lowest_recall_class} observations than of the other classes."
    )

    if lowest_recall_support is not None:
        recall_answer += (
            f" The reported support for this class was "
            f"{lowest_recall_support:,.0f} observations."
        )
else:
    recall_answer = (
        "The class with the lowest recall cannot be identified from "
        "text_metrics.csv because the file does not contain class-level "
        "recall values. Review the printed classification report or save "
        "the per-class metrics during training."
    )

questions_and_answers.append({
    "Question": "Which class is most frequently missed?",
    "Answer": recall_answer
})


# Question 3: Which prediction is least reliable?
if lowest_precision_class is not None:
    precision_answer = (
        f"The least reliable predicted class was "
        f"'{lowest_precision_class}', with a precision of "
        f"{lowest_precision_value:.4f}. A lower precision means that "
        f"a larger proportion of texts predicted as "
        f"{lowest_precision_class} actually belonged to other classes."
    )
else:
    precision_answer = (
        "The least reliable prediction cannot be identified from "
        "text_metrics.csv because class-level precision is unavailable. "
        "Review the classification report and the false positives in the "
        "corresponding confusion-matrix column."
    )

questions_and_answers.append({
    "Question": "Which prediction is least reliable?",
    "Answer": precision_answer
})


# Question 4: Is there class imbalance?
if support_column is not None and class_column is not None:

    support_rows = metrics_df[
        metrics_df[class_column]
        .astype(str)
        .str.lower()
        .isin(
            [
                "negative",
                "neutral",
                "positive"
            ]
        )
    ].copy()

    if not support_rows.empty:
        support_summary = (
            support_rows[
                [
                    class_column,
                    support_column
                ]
            ]
            .drop_duplicates(
                subset=[class_column]
            )
        )

        support_text = ", ".join(
            f"{row[class_column]}: {int(row[support_column]):,}"
            for _, row in support_summary.iterrows()
        )

        imbalance_answer = (
            f"Yes. The class supports were {support_text}. "
            "The classes do not contribute the same number of examples, "
            "so accuracy must not be interpreted alone."
        )
    else:
        imbalance_answer = (
            "Class-level support was not available in the metrics rows. "
            "However, the LAB04 audit already identified severe class "
            "imbalance, especially for neutral sentiment."
        )
else:
    imbalance_answer = (
        "Yes. The LAB04 audit showed that the processed dataset is "
        "imbalanced: positive and negative contain many more observations "
        "than neutral. Therefore, macro F1 and class-level recall are more "
        "informative than accuracy alone."
    )

questions_and_answers.append({
    "Question": "Does the dataset contain class imbalance?",
    "Answer": imbalance_answer
})


# Question 5: Can the model generalize?
generalization_answer = (
    "The result supports limited generalization only to texts similar to "
    "the audited corpus: reconstructed Spanish-language Twitter content "
    "from the documented collection periods and represented topics. "
    "The evaluation does not demonstrate equivalent performance for "
    "current social-media language, formal documents, product reviews, "
    "medical texts, every Spanish-speaking region, or text outside the "
    "training vocabulary."
)

questions_and_answers.append({
    "Question": "Can the selected model generalize?",
    "Answer": generalization_answer
})


# ------------------------------------------------------------------------------
# 12. DISPLAY THE QUESTIONS AND AUTOMATIC ANSWERS
# ------------------------------------------------------------------------------

interpretation_df = pd.DataFrame(
    questions_and_answers
)

print("\n" + "=" * 70)
print("REQUIRED METRIC INTERPRETATION")
print("=" * 70)

display(interpretation_df)


# ------------------------------------------------------------------------------
# 13. DISPLAY THE LARGEST OBSERVED CONFUSION
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("LARGEST OBSERVED CLASSIFICATION ERROR")
print("=" * 70)

if largest_confusion_description is not None:
    print(largest_confusion_description)
else:
    print(
        "A numeric confusion could not be extracted automatically. "
        "Open reports/confusion_text.png and identify the largest "
        "off-diagonal cell."
    )


# ------------------------------------------------------------------------------
# 14. GENERATE AN INTERPRETIVE EXPLANATION TEMPLATE
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("INTERPRETIVE EXPLANATION")
print("=" * 70)

explanation_parts = [
    (
        f"{best_model_name} obtained the highest macro F1-score "
        f"({best_macro_f1:.4f}) among the evaluated pipelines."
    )
]

if baseline_macro_f1 is not None:
    explanation_parts.append(
        f"The baseline obtained a macro F1-score of "
        f"{baseline_macro_f1:.4f}, producing an absolute improvement "
        f"of {macro_f1_improvement:.4f}."
    )

if lowest_recall_class is not None:
    explanation_parts.append(
        f"However, the '{lowest_recall_class}' class presented the "
        f"lowest recall ({lowest_recall_value:.4f}), indicating that "
        f"the classifier missed a comparatively large proportion of "
        f"real {lowest_recall_class} texts."
    )
else:
    explanation_parts.append(
        "The class with the lowest recall must be confirmed using the "
        "per-class classification report."
    )

if largest_confusion_description is not None:
    explanation_parts.append(
        largest_confusion_description
    )

explanation_parts.append(
    "Because the corpus is strongly imbalanced and neutral sentiment is "
    "underrepresented, the model should not be published based on accuracy "
    "alone. The next experiment should prioritize class-level recall, macro "
    "F1-score, duplicate control, and manual review of the dominant error "
    "categories before deployment."
)

# Join and display the explanation.
interpretive_explanation = " ".join(
    explanation_parts
)

print(interpretive_explanation)


# ------------------------------------------------------------------------------
# 15. DISPLAY THE SUPPORTING REPORT FILES
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("SUPPORTING REPORT FILES")
print("=" * 70)
print(f"Metrics file exists:          {METRICS_PATH.exists()}")
print(f"Confusion matrix exists:      {CONFUSION_MATRIX_PATH.exists()}")
print(f"Error analysis file exists:   {ERROR_ANALYSIS_PATH.exists()}")
````

## Confusion Matrix Analysis

The confusion matrix summarizes the predictions generated by the selected text-classification model. The rows represent the **true sentiment labels**, while the columns represent the **predicted sentiment labels**.

### Correct Predictions

The main diagonal represents the observations classified correctly:

- **Negative:** 3,511 texts were correctly classified as negative.
- **Neutral:** 87 texts were correctly classified as neutral.
- **Positive:** 4,594 texts were correctly classified as positive.

The model produced **8,192 correct predictions out of 9,922 test observations**, corresponding to an approximate accuracy of **82.56%**.

However, accuracy alone does not provide a complete evaluation because the three sentiment classes are severely imbalanced.

### Most Frequent Confusion

The largest off-diagonal cell corresponds to:

- **722 positive texts incorrectly classified as negative.**

This is the most frequent individual classification error in the confusion matrix.

In a public-opinion or customer-feedback monitoring system, this error could cause favorable opinions to be interpreted as complaints or dissatisfaction. As a result, the application could overestimate negative sentiment, underestimate positive sentiment, and present a more negative view of the analyzed population than the data actually supports.

The second-largest error corresponds to:

- **342 negative texts incorrectly classified as positive.**

This type of error could hide complaints or unfavorable opinions by presenting them as positive feedback.

### Class With the Greatest Difficulty

Although the most frequent individual error was the classification of positive texts as negative, the **neutral class presented the greatest proportional difficulty**.

The recall for each class was approximately:

- **Negative recall:** 86.71%
- **Neutral recall:** 23.39%
- **Positive recall:** 83.51%

Only **87 of the 372 truly neutral texts** were correctly classified. The remaining neutral texts were classified as follows:

- 197 neutral texts were classified as negative.
- 88 neutral texts were classified as positive.

Therefore, **285 of the 372 neutral observations were misclassified**.

This means that the model correctly recovered only approximately **23 out of every 100 neutral texts**.

### Least Reliable Prediction

The neutral class also had the lowest approximate precision.

The model predicted the neutral class a total of 468 times:

- 196 negative texts were predicted as neutral.
- 87 neutral texts were predicted correctly.
- 185 positive texts were predicted as neutral.

Only 87 of these 468 neutral predictions were correct, corresponding to an approximate neutral precision of **18.59%**.

Therefore, when the model predicts that a text is neutral, that prediction is correct in only approximately **19 out of every 100 cases**.

### Relationship With Class Imbalance

The poor performance of the neutral class is consistent with the class imbalance identified during the corpus audit:

- Positive: 55.79%
- Negative: 40.50%
- Neutral: 3.71%

Because neutral sentiment represents only a small proportion of the training corpus, the model has considerably fewer neutral examples from which to learn.

For this reason, the global accuracy of 82.56% should not be interpreted as evidence that the model performs equally well across all three classes.

### Practical Interpretation

The selected model performs relatively well when distinguishing positive and negative sentiment, but it does not reliably recognize neutral sentiment.

The most frequent individual confusion was **positive → negative**, with 722 errors. This error could inflate the apparent amount of negative sentiment and distort conclusions about satisfaction, approval, or public opinion.

The most serious class-level limitation was the neutral class, which had an approximate recall of 23.39% and precision of 18.59%. This indicates that the model frequently assigns neutral texts to one of the two dominant sentiment classes and that most neutral predictions are incorrect.

### Decision Before Publication

The model should not be published based only on its global accuracy.

Before deployment, the next experiment should:

1. Review duplicated texts before creating the final training and test partitions.
2. Investigate neutral examples manually.
3. Categorize errors involving negation, irony, ambiguity, dialect, insufficient text, and questionable labels.
4. Evaluate class-sensitive training strategies.
5. Compare macro F1-score and recall by class.
6. Report the confusion matrix together with the global metrics.
7. Clearly warn users that neutral predictions are substantially less reliable.

### Final Interpretation

The selected classifier achieved a global accuracy of approximately **82.56%**. However, the confusion matrix shows that this result is driven mainly by the strong performance on the two majority classes.

The largest off-diagonal error consisted of **722 positive texts classified as negative**. In the intended context, this error could cause favorable opinions to be interpreted as dissatisfaction and artificially increase the estimated amount of negative sentiment.

The **neutral class presented the greatest difficulty**, with an approximate recall of **23.39%** and precision of **18.59%**. This limitation is consistent with the severe underrepresentation of neutral sentiment in the audited corpus.

Therefore, the model requires additional evaluation and improvement before publication, particularly for the neutral class.

````python
# ==============================================================================
# LAB05 — CELL 10: RUN CORRECTED APPLICATION FUNCTIONAL TESTS
# Objective: Verify that the saved pipeline generates valid predictions,
# calculates probabilities correctly, and identifies uncertain predictions
# without treating model outputs as objective truths.
# ==============================================================================

# Import Path for operating-system-independent path management.
from pathlib import Path

# Import Joblib to load the saved text-classification pipeline.
import joblib

# Import pandas to organize and display the functional test results.
import pandas as pd


# ------------------------------------------------------------------------------
# 1. DEFINE THE PROJECT AND MODEL PATHS
# ------------------------------------------------------------------------------

# Define the root directory of the NLP sentiment-analysis project.
PROJECT_PATH = Path(
    "/content/INF-8239-Ciencia-de-Datos-II/"
    "Unit_02_Natural_Language_Processing/"
    "nlp_sentiment_analysis"
)

# Define the saved text-classification pipeline.
MODEL_PATH = (
    PROJECT_PATH
    / "models"
    / "text_model.joblib"
)

# Define the Streamlit application file.
APP_PATH = (
    PROJECT_PATH
    / "app"
    / "streamlit_app.py"
)


# ------------------------------------------------------------------------------
# 2. VERIFY AND LOAD THE SAVED MODEL
# ------------------------------------------------------------------------------

# Stop execution if the saved pipeline is unavailable.
if not MODEL_PATH.exists():
    raise FileNotFoundError(
        f"The saved text model was not found:\n{MODEL_PATH}"
    )

# Load the complete TF-IDF and classifier pipeline.
text_model = joblib.load(MODEL_PATH)

# Verify that the selected classifier supports probability estimates.
if not hasattr(text_model, "predict_proba"):
    raise AttributeError(
        "The saved model does not support predict_proba(). "
        "Probability-based uncertainty cannot be calculated."
    )

# Retrieve the sentiment classes known by the trained pipeline.
model_classes = text_model.classes_.tolist()


# ------------------------------------------------------------------------------
# 3. DEFINE THE UNCERTAINTY RULE
# ------------------------------------------------------------------------------

# Define the uncertainty threshold used by the application.
WARNING_THRESHOLD = 0.35

# Define a small tolerance for floating-point comparisons.
FLOAT_TOLERANCE = 1e-9


# ------------------------------------------------------------------------------
# 4. DEFINE THE FUNCTIONAL TEST CASES
# ------------------------------------------------------------------------------

# Define clear, ambiguous, potentially neutral, and out-of-domain examples.
test_cases = [
    {
        "description": "Clearly positive example",
        "text": "Excelente servicio, estoy muy satisfecho.",
        "expected_class": "positive"
    },
    {
        "description": "Clearly negative example",
        "text": "El servicio fue terrible y no lo recomiendo.",
        "expected_class": "negative"
    },
    {
        "description": "Ambiguous example",
        "text": "Bueno, pudo haber sido peor.",
        "expected_class": None
    },
    {
        "description": "Potentially neutral example",
        "text": "La reunión comienza mañana a las diez.",
        "expected_class": None
    },
    {
        "description": "Out-of-domain example",
        "text": "La válvula neumática opera a una presión de seis bares.",
        "expected_class": None
    }
]


# ------------------------------------------------------------------------------
# 5. GENERATE AND EVALUATE THE PREDICTIONS
# ------------------------------------------------------------------------------

# Create a list to store the results of every functional test.
functional_results = []

# Evaluate each example independently.
for test_case in test_cases:

    # Store the current input text.
    input_text = test_case["text"]

    # Generate the predicted sentiment class.
    predicted_class = text_model.predict(
        [input_text]
    )[0]

    # Generate the probability assigned to each sentiment class.
    predicted_probabilities = text_model.predict_proba(
        [input_text]
    )[0]

    # Retrieve the highest predicted probability.
    maximum_probability = float(
        predicted_probabilities.max()
    )

    # Calculate uncertainty as one minus the maximum probability.
    uncertainty = 1.0 - maximum_probability

    # Trigger the warning when uncertainty reaches or exceeds the threshold.
    warning_triggered = (
        uncertainty
        >= WARNING_THRESHOLD - FLOAT_TOLERANCE
    )

    # Confirm that the prediction belongs to the trained class vocabulary.
    valid_class_check = predicted_class in model_classes

    # Confirm that the probabilities sum approximately to one.
    probability_sum_check = abs(
        float(predicted_probabilities.sum()) - 1.0
    ) < FLOAT_TOLERANCE

    # Compare the prediction only for examples with an expected clear class.
    if test_case["expected_class"] is None:
        expected_class_match = "Not required"
    else:
        expected_class_match = (
            predicted_class
            == test_case["expected_class"]
        )

    # Create a readable probability summary.
    probability_summary = {
        class_name: round(float(class_probability), 4)
        for class_name, class_probability in zip(
            model_classes,
            predicted_probabilities
        )
    }

    # Store the complete functional result.
    functional_results.append({
        "description": test_case["description"],
        "text": input_text,
        "expected_class": (
            test_case["expected_class"]
            if test_case["expected_class"] is not None
            else "No mandatory class"
        ),
        "prediction": predicted_class,
        "maximum_probability": maximum_probability,
        "uncertainty": uncertainty,
        "warning_triggered": warning_triggered,
        "valid_class": valid_class_check,
        "probabilities_sum_to_one": probability_sum_check,
        "expected_class_match": expected_class_match,
        "class_probabilities": probability_summary
    })


# ------------------------------------------------------------------------------
# 6. DISPLAY THE FUNCTIONAL TEST RESULTS
# ------------------------------------------------------------------------------

# Convert the results into a readable DataFrame.
functional_results_df = pd.DataFrame(
    functional_results
)

print("=" * 70)
print("STREAMLIT MODEL FUNCTIONAL TESTS")
print("=" * 70)
print(f"Model path: {MODEL_PATH}")
print(f"Model classes: {model_classes}")
print(f"Uncertainty threshold: {WARNING_THRESHOLD:.2f}")

# Display the main functional results.
display(
    functional_results_df[
        [
            "description",
            "text",
            "expected_class",
            "prediction",
            "maximum_probability",
            "uncertainty",
            "warning_triggered",
            "valid_class",
            "probabilities_sum_to_one",
            "expected_class_match"
        ]
    ].round({
        "maximum_probability": 4,
        "uncertainty": 4
    })
)


# ------------------------------------------------------------------------------
# 7. DISPLAY THE CLASS PROBABILITIES
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("CLASS PROBABILITIES")
print("=" * 70)

# Display the probability assigned to every class for each example.
for result in functional_results:
    print(f"\n{result['description']}")
    print(f"Text: {result['text']}")
    print(f"Prediction: {result['prediction']}")
    print(f"Probabilities: {result['class_probabilities']}")


# ------------------------------------------------------------------------------
# 8. VERIFY THE INTERFACE SCOPE WARNING
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("INTERFACE SCOPE WARNING")
print("=" * 70)

# Verify that the Streamlit application exists.
if not APP_PATH.exists():
    print(f"The Streamlit application was not found:\n{APP_PATH}")

else:
    # Read the Streamlit application source code.
    app_content = APP_PATH.read_text(
        encoding="utf-8"
    ).lower()

    # Define warning-related concepts expected in the interface.
    warning_keywords = [
        "prediction",
        "warning",
        "limitation",
        "scope",
        "corpus",
        "not an objective fact",
        "not a fact"
    ]

    # Identify the warning terms present in the application.
    detected_keywords = [
        keyword
        for keyword in warning_keywords
        if keyword in app_content
    ]

    if detected_keywords:
        print("Scope-warning language was detected in the application:")

        for keyword in detected_keywords:
            print(f"- {keyword}")

    else:
        print(
            "Warning: no clear scope-warning language was detected. "
            "The interface should explain that the output is a model "
            "prediction conditioned by the training corpus."
        )


# ------------------------------------------------------------------------------
# 9. DISPLAY THE FINAL FUNCTIONAL INTERPRETATION
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("FUNCTIONAL TEST INTERPRETATION")
print("=" * 70)

print(
    "Clear positive and negative examples are compared with an expected "
    "class. Ambiguous, potentially neutral, and out-of-domain examples "
    "do not receive a mandatory expected class because the model output "
    "must be preserved and interpreted honestly."
)

print(
    "\nThe low-confidence warning is activated when uncertainty reaches "
    "or exceeds the configured threshold."
)

print(
    "\nA successful functional test confirms that the saved pipeline can "
    "generate valid predictions and probabilities. It does not prove that "
    "every prediction is correct or that the model generalizes beyond the "
    "audited corpus."
)
````

# Interpretive Closing

## Main Result

Exercise 03 successfully integrated the corpus-auditing workflow from U02.LAB04 with the text-classification workflow from U02.LAB05. The approved Spanish-language corpus was downloaded, reconstructed from its encoded representation, audited, processed, and used to train three comparable classification pipelines.

The final processed dataset contained 40,045 observations distributed across three sentiment classes: 22,340 positive texts, 16,219 negative texts, and 1,486 neutral texts. The workflow compared a `DummyClassifier` baseline, Complement Naive Bayes, and Logistic Regression using TF-IDF features. Macro F1-score was prioritized because the classes were highly imbalanced.

## Selected Model and Supporting Evidence

The model with the highest macro F1-score was selected and saved as a reusable pipeline in `models/text_model.joblib`. The saved object preserves both the TF-IDF vectorizer and the trained classifier, allowing new text to be transformed and classified through a single pipeline.

The selected model produced 8,192 correct predictions from 9,922 test observations, corresponding to an approximate accuracy of 82.56%. However, model selection and interpretation were not based on accuracy alone. Macro F1-score, precision, recall, class support, and the confusion matrix were also examined.

The model file was successfully loaded with Joblib, produced predictions for new Spanish texts, and passed the automated model tests. The Streamlit application also started successfully in a newly created isolated environment containing only the dependencies exported to `requirements-cloud.txt`.

## Class With the Greatest Difficulty

Neutral sentiment was the most difficult class. Only 87 of 372 truly neutral observations were classified correctly, producing an approximate recall of 23.39%. The model predicted the neutral class 468 times, but only 87 predictions were correct, corresponding to an approximate precision of 18.59%.

This limitation is consistent with the corpus distribution because neutral sentiment represented only 3.71% of the processed observations.

## Most Frequent Error Type

The largest numerical confusion was **positive sentiment predicted as negative**, with 722 occurrences. This is different from the linguistic error category required by the manual review.

**Manual category to complete after reviewing at least 20 errors:** `[negation / irony / ambiguity / dialect / insufficient text / questionable label / out of domain]`.

The original labels and model predictions must remain unchanged during this categorization.

## Contextual Impact

Classifying positive texts as negative could overestimate dissatisfaction and produce an unnecessarily negative interpretation of public opinion. The poor recognition of neutral sentiment could also force factual or ambiguous messages into positive or negative categories, creating an exaggerated impression of polarization.

## Dataset Limitation

The principal limitation is severe class imbalance. Other limitations include historical Twitter language, reconstructed rather than original text formatting, duplicated observations, informal expressions, possible sarcasm, and limited representation of current Spanish usage and regional varieties.

## Decision Before Deployment

The application is technically portable and operational, but the model should be presented as an experimental classifier rather than an authoritative sentiment evaluator. Before deployment, duplicate leakage should be controlled, at least 20 errors should be categorized, and a new experiment should address the dominant error category and the weak neutral-class performance.

````python
# ==============================================================================
# AUTOMATIC COMMIT AND AUTHENTICATED PUSH
# Objective: Preserve empty directories, stage repository changes, create a
# commit when needed, push the current branch securely, and verify Git status.
# ==============================================================================

# Import os for environment and file-system operations.
import os

# Import stat to configure the temporary authentication script permissions.
import stat

# Import subprocess to execute Git commands and verify their results.
import subprocess

# Import Path for operating-system-independent path management.
from pathlib import Path

# Import userdata to retrieve the GitHub token from Colab Secrets.
from google.colab import userdata


# ------------------------------------------------------------------------------
# 1. DEFINE THE REPOSITORY CONFIGURATION
# ------------------------------------------------------------------------------

# Define the repository root directory.
REPO_PATH = Path(
    "/content/INF-8239-Ciencia-de-Datos-II"
)

# Define the Unit 02 directory.
UNIT_02_PATH = (
    REPO_PATH
    / "Unit_02_Natural_Language_Processing"
)

# Define the GitHub repository information.
GITHUB_USER = "pascual-francisco"
GITHUB_REPO = "INF-8239-Ciencia-de-Datos-II"

# Define the commit message.
COMMIT_MESSAGE = "chore: sync repository updates"

# Retrieve the GitHub token from Colab Secrets.
GITHUB_TOKEN = userdata.get("TOKEN")

# Define the clean GitHub remote URL.
CLEAN_REMOTE_URL = (
    f"https://github.com/{GITHUB_USER}/{GITHUB_REPO}.git"
)


# ------------------------------------------------------------------------------
# 2. VALIDATE THE REPOSITORY AND TOKEN
# ------------------------------------------------------------------------------

# Confirm that the repository directory exists.
if not REPO_PATH.exists():
    raise FileNotFoundError(
        f"The repository was not found:\n{REPO_PATH}"
    )

# Confirm that the repository contains a .git directory.
if not (REPO_PATH / ".git").exists():
    raise FileNotFoundError(
        f"The directory is not a Git repository:\n{REPO_PATH}"
    )

# Confirm that the GitHub token is available.
if not GITHUB_TOKEN:
    raise ValueError(
        "The GitHub token was not found. "
        "Create a Colab Secret named TOKEN and enable notebook access."
    )


# ------------------------------------------------------------------------------
# 3. DEFINE A REUSABLE GIT COMMAND FUNCTION
# ------------------------------------------------------------------------------

def run_git(
    arguments: list[str],
    *,
    check: bool = True,
    capture_output: bool = False,
    environment: dict[str, str] | None = None
) -> subprocess.CompletedProcess:
    """
    Execute a Git command from the repository root.

    Parameters
    ----------
    arguments:
        Git arguments without the initial 'git'.
    check:
        Raise an error when the command returns a nonzero exit code.
    capture_output:
        Capture stdout and stderr instead of displaying them directly.
    environment:
        Optional environment variables used by the Git command.
    """

    return subprocess.run(
        ["git", *arguments],
        cwd=REPO_PATH,
        check=check,
        text=True,
        capture_output=capture_output,
        env=environment
    )


# ------------------------------------------------------------------------------
# 4. PREPARE EMPTY DIRECTORIES
# ------------------------------------------------------------------------------

print("=" * 70)
print("PREPARING EMPTY DIRECTORIES")
print("=" * 70)

if UNIT_02_PATH.exists():

    created_gitkeep_files = []

    # Walk through every directory inside Unit 02.
    for directory_path, directory_names, file_names in os.walk(
        UNIT_02_PATH
    ):
        # A directory is empty when it has no folders and no files.
        if not directory_names and not file_names:
            gitkeep_path = Path(directory_path) / ".gitkeep"
            gitkeep_path.touch()
            created_gitkeep_files.append(gitkeep_path)

    if created_gitkeep_files:
        print(
            f"Created {len(created_gitkeep_files)} "
            ".gitkeep file(s)."
        )
    else:
        print("No empty directories required a .gitkeep file.")

else:
    raise FileNotFoundError(
        f"The Unit 02 directory was not found:\n{UNIT_02_PATH}"
    )


# ------------------------------------------------------------------------------
# 5. DISPLAY THE CURRENT CHANGES
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("CHANGES DETECTED")
print("=" * 70)

# Display modified, deleted, renamed, and untracked files.
run_git([
    "status",
    "--short",
    "--untracked-files=all"
])


# ------------------------------------------------------------------------------
# 6. STAGE THE REPOSITORY CHANGES
# ------------------------------------------------------------------------------

# Stage all changes that are not excluded by .gitignore.
run_git([
    "add",
    "-A"
])

print("\n" + "=" * 70)
print("FILES STAGED FOR COMMIT")
print("=" * 70)

# Retrieve the exact list of staged changes.
staged_result = run_git(
    [
        "diff",
        "--cached",
        "--name-status"
    ],
    capture_output=True
)

staged_output = staged_result.stdout.strip()

# Display the staged changes.
if staged_output:
    print(staged_output)
else:
    print("No files are currently staged.")


# ------------------------------------------------------------------------------
# 7. CREATE A COMMIT ONLY WHEN CHANGES EXIST
# ------------------------------------------------------------------------------

if staged_output:
    print("\nCreating commit...")

    run_git([
        "commit",
        "-m",
        COMMIT_MESSAGE
    ])

    print(f"Commit created: {COMMIT_MESSAGE}")

else:
    print("\nNo new changes to commit.")


# ------------------------------------------------------------------------------
# 8. IDENTIFY THE CURRENT BRANCH
# ------------------------------------------------------------------------------

branch_result = run_git(
    [
        "branch",
        "--show-current"
    ],
    capture_output=True
)

current_branch = branch_result.stdout.strip()

if not current_branch:
    raise RuntimeError(
        "Git could not determine the current branch."
    )


# ------------------------------------------------------------------------------
# 9. CONFIGURE A CLEAN REMOTE URL
# ------------------------------------------------------------------------------

# Keep the permanent origin URL free from credentials.
run_git([
    "remote",
    "set-url",
    "origin",
    CLEAN_REMOTE_URL
])


# ------------------------------------------------------------------------------
# 10. CREATE TEMPORARY NON-INTERACTIVE AUTHENTICATION
# ------------------------------------------------------------------------------

# Define a temporary Git authentication helper.
askpass_path = Path("/tmp/github_colab_askpass.sh")

# Create a temporary script that responds to Git credential prompts.
askpass_content = """#!/bin/sh
case "$1" in
    *Username*) printf '%s\\n' "$GITHUB_USERNAME" ;;
    *Password*) printf '%s\\n' "$GITHUB_TOKEN" ;;
    *) printf '\\n' ;;
esac
"""

# Save the authentication helper.
askpass_path.write_text(
    askpass_content,
    encoding="utf-8"
)

# Allow only the current user to execute or modify the helper.
askpass_path.chmod(
    stat.S_IRUSR
    | stat.S_IWUSR
    | stat.S_IXUSR
)

# Create an environment for the authenticated Git push.
push_environment = os.environ.copy()

# Tell Git which temporary authentication helper to use.
push_environment["GIT_ASKPASS"] = str(askpass_path)

# Prevent Git from opening an interactive credential prompt.
push_environment["GIT_TERMINAL_PROMPT"] = "0"

# Supply the GitHub username only through the process environment.
push_environment["GITHUB_USERNAME"] = GITHUB_USER

# Supply the GitHub token only through the process environment.
push_environment["GITHUB_TOKEN"] = GITHUB_TOKEN


# ------------------------------------------------------------------------------
# 11. PUSH THE CURRENT BRANCH
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("PUSHING TO GITHUB")
print("=" * 70)
print(f"Current branch: {current_branch}")

try:
    # Push the current branch and establish its upstream relationship.
    run_git(
        [
            "push",
            "--set-upstream",
            "origin",
            current_branch
        ],
        environment=push_environment
    )

finally:
    # Delete the temporary authentication helper after the push attempt.
    if askpass_path.exists():
        askpass_path.unlink()


# ------------------------------------------------------------------------------
# 12. DISPLAY THE LATEST COMMITS
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("LATEST COMMITS")
print("=" * 70)

# Display the five most recent commits.
run_git([
    "log",
    "--oneline",
    "--decorate",
    "-5"
])


# ------------------------------------------------------------------------------
# 13. VERIFY THE FINAL SYNCHRONIZATION STATUS
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SYNCHRONIZATION STATUS")
print("=" * 70)

# Display the final branch and working-tree status.
run_git([
    "status",
    "--short",
    "--branch"
])

print("\nRepository synchronized successfully.")
````
