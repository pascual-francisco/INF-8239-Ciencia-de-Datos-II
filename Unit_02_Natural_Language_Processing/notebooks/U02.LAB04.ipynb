```python
# ------------------------------------------------------------------------------
# Create the Unit 02 folder structure, .gitkeep files, and notebooks
# ------------------------------------------------------------------------------

from pathlib import Path
import json

# Repository root
repo_dir = Path("/content/INF-8239-Ciencia-de-Datos-II")

# Unit 02 root directory
unit_dir = repo_dir / "Unit_02_Natural_Language_Processing"

# ------------------------------------------------------------------------------
# 1. Define the folder structure
# ------------------------------------------------------------------------------

folders = [
    unit_dir / "app",
    unit_dir / "data" / "raw",
    unit_dir / "data" / "processed",
    unit_dir / "docs",
    unit_dir / "notebooks",
    unit_dir / "reports" / "figures",
    unit_dir / "scripts",
    unit_dir / "src",
    unit_dir / "tests",
]

# Create folders without modifying existing content
for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------------------------
# 2. Add .gitkeep files to empty folders
# ------------------------------------------------------------------------------

# Git does not track empty folders.
# These files allow the complete folder structure to appear on GitHub.
gitkeep_folders = [
    unit_dir / "app",
    unit_dir / "data" / "raw",
    unit_dir / "data" / "processed",
    unit_dir / "docs",
    unit_dir / "reports" / "figures",
    unit_dir / "scripts",
    unit_dir / "src",
    unit_dir / "tests",
]

print("=" * 70)
print("CREATING UNIT 02 PROJECT STRUCTURE")
print("=" * 70)

for folder in gitkeep_folders:
    gitkeep_path = folder / ".gitkeep"

    if gitkeep_path.exists():
        print(
            f"[SKIPPED] .gitkeep already exists: "
            f"{gitkeep_path.relative_to(repo_dir)}"
        )
    else:
        gitkeep_path.touch()
        print(
            f"[CREATED] .gitkeep: "
            f"{gitkeep_path.relative_to(repo_dir)}"
        )

# ------------------------------------------------------------------------------
# 3. Define the Unit 02 notebooks
# ------------------------------------------------------------------------------

notebooks_dir = unit_dir / "notebooks"

notebook_names = [
    "U02.LAB04.ipynb",
    "U02.LAB05.ipynb",
    "U02.LAB06.ipynb",
    "U02.LAB07.ipynb",
    "U02.Exercise_03.ipynb",
    "U02.Exercise_04.ipynb",
]

# Minimal valid Jupyter Notebook structure
empty_notebook = {
    "cells": [],
    "metadata": {
        "kernelspec": {
            "display_name": "Python 3",
            "language": "python",
            "name": "python3"
        },
        "language_info": {
            "name": "python",
            "version": "3"
        }
    },
    "nbformat": 4,
    "nbformat_minor": 5
}

# ------------------------------------------------------------------------------
# 4. Create the notebooks without overwriting existing files
# ------------------------------------------------------------------------------

for notebook_name in notebook_names:
    notebook_path = notebooks_dir / notebook_name

    if notebook_path.exists():
        print(f"[SKIPPED] Notebook already exists: {notebook_name}")
        continue

    with notebook_path.open("w", encoding="utf-8") as file:
        json.dump(
            empty_notebook,
            file,
            ensure_ascii=False,
            indent=2
        )

    print(f"[CREATED] Notebook: {notebook_name}")

# ------------------------------------------------------------------------------
# 5. Verify the complete folder structure
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("VERIFYING UNIT 02 FOLDER STRUCTURE")
print("=" * 70)

all_correct = True

for folder in folders:
    relative_folder = folder.relative_to(repo_dir)

    if folder.is_dir():
        print(f"[OK]      Folder: {relative_folder}")
    else:
        print(f"[MISSING] Folder: {relative_folder}")
        all_correct = False

# ------------------------------------------------------------------------------
# 6. Verify the .gitkeep files
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("VERIFYING .GITKEEP FILES")
print("=" * 70)

for folder in gitkeep_folders:
    gitkeep_path = folder / ".gitkeep"
    relative_path = gitkeep_path.relative_to(repo_dir)

    if gitkeep_path.is_file():
        print(f"[OK]      File: {relative_path}")
    else:
        print(f"[MISSING] File: {relative_path}")
        all_correct = False

# ------------------------------------------------------------------------------
# 7. Verify the notebooks
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("VERIFYING UNIT 02 NOTEBOOKS")
print("=" * 70)

for notebook_name in notebook_names:
    notebook_path = notebooks_dir / notebook_name
    relative_path = notebook_path.relative_to(repo_dir)

    if notebook_path.is_file():
        print(f"[OK]      Notebook: {relative_path}")
    else:
        print(f"[MISSING] Notebook: {relative_path}")
        all_correct = False

# ------------------------------------------------------------------------------
# 8. Display the final verification result
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)

if all_correct:
    print("RESULT: Unit 02 structure created and verified successfully.")
else:
    print("RESULT: The Unit 02 structure requires review.")

print("=" * 70)
```
