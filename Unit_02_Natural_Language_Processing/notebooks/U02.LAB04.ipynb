```python
# ==============================================================================
# CREATE A REPRODUCIBLE UNIT STRUCTURE
#
# Purpose:
# 1. Create the standard folder structure for a course unit.
# 2. Add .gitkeep files so Git can track empty directories.
# 3. Create valid empty Jupyter notebooks.
# 4. Avoid overwriting folders or notebooks that already exist.
# 5. Verify that the complete structure was created correctly.
#
# Reuse:
# For future units, modify only the CONFIGURATION section below.
# ==============================================================================

from pathlib import Path
import json


# ==============================================================================
# CONFIGURATION
#
# Update these variables when creating a new unit or topic.
# ==============================================================================

# Absolute path to the cloned GitHub repository in Google Colab
REPO_DIR = Path("/content/INF-8239-Ciencia-de-Datos-II")

# Name of the unit directory to create
UNIT_NAME = "Unit_02_Natural_Language_Processing"

# Names of the notebooks required for this unit
NOTEBOOK_NAMES = [
    "U02.LAB04.ipynb",
    "U02.LAB05.ipynb",
    "U02.LAB06.ipynb",
    "U02.LAB07.ipynb",
    "U02.Exercise_03.ipynb",
    "U02.Exercise_04.ipynb",
]


# ==============================================================================
# 1. VALIDATE THE REPOSITORY LOCATION
# ==============================================================================

# Confirm that the cloned repository exists before creating any content
if not REPO_DIR.is_dir():
    raise FileNotFoundError(
        f"The repository directory does not exist:\n{REPO_DIR}"
    )

# Build the complete path for the selected unit
unit_dir = REPO_DIR / UNIT_NAME


# ==============================================================================
# 2. DEFINE THE STANDARD FOLDER STRUCTURE
#
# This structure separates raw data, processed data, documentation, notebooks,
# reports, reusable source code, executable scripts, tests, and applications.
# ==============================================================================

folders = [
    unit_dir / "app",
    unit_dir / "data" / "raw",
    unit_dir / "data" / "processed",
    unit_dir / "docs",
    unit_dir / "notebooks",
    unit_dir / "reports" / "figures",
    unit_dir / "scripts",
    unit_dir / "src",
    unit_dir / "tests",
]


# ==============================================================================
# 3. CREATE THE FOLDERS
#
# parents=True creates any missing parent directories.
# exist_ok=True prevents errors when a directory already exists.
# Existing folders and their contents are not deleted or modified.
# ==============================================================================

print("=" * 78)
print("CREATING PROJECT STRUCTURE")
print("=" * 78)

for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)
    print(f"[READY]   {folder.relative_to(REPO_DIR)}")


# ==============================================================================
# 4. ADD .GITKEEP FILES
#
# Git tracks files, not empty directories. A .gitkeep file allows empty project
# directories to be included in commits and displayed on GitHub.
#
# The notebooks directory does not require .gitkeep because it will contain
# actual Jupyter Notebook files.
# ==============================================================================

gitkeep_folders = [
    unit_dir / "app",
    unit_dir / "data" / "raw",
    unit_dir / "data" / "processed",
    unit_dir / "docs",
    unit_dir / "reports" / "figures",
    unit_dir / "scripts",
    unit_dir / "src",
    unit_dir / "tests",
]

print("\n" + "=" * 78)
print("CREATING .GITKEEP FILES")
print("=" * 78)

for folder in gitkeep_folders:
    gitkeep_path = folder / ".gitkeep"
    relative_path = gitkeep_path.relative_to(REPO_DIR)

    # Do not recreate the file if it already exists
    if gitkeep_path.exists():
        print(f"[SKIPPED] {relative_path}")
    else:
        gitkeep_path.touch()
        print(f"[CREATED] {relative_path}")


# ==============================================================================
# 5. DEFINE A MINIMAL VALID JUPYTER NOTEBOOK
#
# Every .ipynb file is a JSON document. This dictionary contains the minimum
# structure needed for Jupyter or Google Colab to open the files correctly.
# ==============================================================================

empty_notebook = {
    "cells": [],
    "metadata": {
        "kernelspec": {
            "display_name": "Python 3",
            "language": "python",
            "name": "python3",
        },
        "language_info": {
            "name": "python",
            "version": "3",
        },
    },
    "nbformat": 4,
    "nbformat_minor": 5,
}


# ==============================================================================
# 6. CREATE THE NOTEBOOKS
#
# Each notebook is created only when it does not already exist.
# This protects completed or partially completed notebooks from being replaced.
# ==============================================================================

notebooks_dir = unit_dir / "notebooks"

print("\n" + "=" * 78)
print("CREATING NOTEBOOKS")
print("=" * 78)

for notebook_name in NOTEBOOK_NAMES:
    notebook_path = notebooks_dir / notebook_name
    relative_path = notebook_path.relative_to(REPO_DIR)

    if notebook_path.exists():
        print(f"[SKIPPED] Existing notebook: {relative_path}")
        continue

    with notebook_path.open("w", encoding="utf-8") as file:
        json.dump(
            empty_notebook,
            file,
            ensure_ascii=False,
            indent=2,
        )

    print(f"[CREATED] Notebook: {relative_path}")


# ==============================================================================
# 7. VERIFY THE FOLDER STRUCTURE
# ==============================================================================

print("\n" + "=" * 78)
print("VERIFYING FOLDER STRUCTURE")
print("=" * 78)

all_correct = True

for folder in folders:
    relative_path = folder.relative_to(REPO_DIR)

    if folder.is_dir():
        print(f"[OK]      Folder: {relative_path}")
    else:
        print(f"[MISSING] Folder: {relative_path}")
        all_correct = False


# ==============================================================================
# 8. VERIFY THE .GITKEEP FILES
# ==============================================================================

print("\n" + "=" * 78)
print("VERIFYING .GITKEEP FILES")
print("=" * 78)

for folder in gitkeep_folders:
    gitkeep_path = folder / ".gitkeep"
    relative_path = gitkeep_path.relative_to(REPO_DIR)

    if gitkeep_path.is_file():
        print(f"[OK]      File: {relative_path}")
    else:
        print(f"[MISSING] File: {relative_path}")
        all_correct = False


# ==============================================================================
# 9. VERIFY THE JUPYTER NOTEBOOKS
# ==============================================================================

print("\n" + "=" * 78)
print("VERIFYING NOTEBOOKS")
print("=" * 78)

for notebook_name in NOTEBOOK_NAMES:
    notebook_path = notebooks_dir / notebook_name
    relative_path = notebook_path.relative_to(REPO_DIR)

    if notebook_path.is_file():
        print(f"[OK]      Notebook: {relative_path}")
    else:
        print(f"[MISSING] Notebook: {relative_path}")
        all_correct = False


# ==============================================================================
# 10. DISPLAY THE FINAL RESULT
# ==============================================================================

print("\n" + "=" * 78)

if all_correct:
    print(f"RESULT: {UNIT_NAME} was created and verified successfully.")
else:
    print(f"RESULT: {UNIT_NAME} requires review.")

print("=" * 78)
```
