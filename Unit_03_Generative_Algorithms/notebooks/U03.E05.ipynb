# 4. Interpretar la factorización

## Resultados obtenidos

- RMSE: 1.0264
- HitRate@10: 0.0375
- Cobertura: 0.0775
- Factores latentes: 20
- Épocas: 12

## Observación

El modelo colaborativo basado en factorización matricial logró predecir ratings retenidos con un RMSE de 1.0264. Adicionalmente, alcanzó un HitRate@10 de 3.75% y una cobertura del catálogo de 7.75%.

## Interpretación del RMSE

El RMSE obtenido indica que el modelo es capaz de aproximar las valoraciones observadas utilizando patrones aprendidos a partir de las interacciones entre usuarios y películas. Un valor menor de RMSE representa una aproximación más precisa de los ratings retenidos.

Sin embargo, un buen RMSE no garantiza automáticamente recomendaciones útiles para el usuario final, por lo que también deben evaluarse métricas de ranking.

## Interpretación del HitRate@10

El HitRate@10 obtenido fue de 3.75%, lo que indica que la película retenida apareció dentro de las diez recomendaciones para una proporción limitada de usuarios.

Este resultado evidencia que la capacidad de predicción de ratings y la calidad del ranking deben evaluarse de forma independiente.

## Interpretación de la cobertura

La cobertura alcanzó 7.75% del catálogo disponible. Esto indica que el modelo utiliza una parte limitada de las películas disponibles para generar recomendaciones.

Una cobertura reducida puede producir recomendaciones más repetitivas y menor diversidad del catálogo mostrado a los usuarios.

## Usuario fuera del entrenamiento

Cuando un usuario no dispone de historial previo, la señal colaborativa no puede calcularse porque no existen interacciones para construir su representación latente.

En estos casos el sistema utiliza un mecanismo de fallback basado en popularidad, permitiendo generar recomendaciones aunque no sean personalizadas.

# 8. Revisar riesgos

## Popularidad dominante

### Observación

Los sistemas recomendadores pueden tender a mostrar repetidamente los mismos elementos populares a una gran cantidad de usuarios.

### Riesgo

La concentración de recomendaciones en un conjunto reducido de películas limita la exposición al resto del catálogo y puede reducir la diversidad de contenido consumido.

### Evidencia en el laboratorio

La cobertura obtenida fue inferior al 11% del catálogo, lo que indica que una parte importante de las películas disponibles nunca aparece en las recomendaciones generadas.

### Posible mitigación

- Incorporar diversidad en el ranking.
- Utilizar sistemas híbridos.
- Penalizar recomendaciones excesivamente populares.

---

## Burbuja de filtro

### Observación

Las recomendaciones pueden concentrarse en una temática muy específica asociada a los gustos históricos del usuario.

### Riesgo

El usuario recibe constantemente contenido similar y disminuye la probabilidad de descubrir nuevas categorías o intereses.

### Evidencia en el laboratorio

En LAB08 se observó que las recomendaciones para Toy Story (1995) compartían prácticamente los mismos géneros, generando una lista altamente homogénea.

### Posible mitigación

- Introducir diversidad en el Top‑N.
- Combinar contenido, colaboración y popularidad.
- Incorporar mecanismos de exploración.

---

## Sesgo de selección

### Observación

El modelo únicamente observa películas que los usuarios decidieron consumir y valorar.

### Riesgo

Las preferencias se estiman utilizando información incompleta, ya que las películas no visualizadas no necesariamente representan desinterés por parte del usuario.

### Evidencia en el laboratorio

MovieLens registra únicamente interacciones observadas mediante ratings, por lo que no existe información sobre contenido que los usuarios nunca llegaron a consumir.

### Posible mitigación

- Incorporar datos implícitos.
- Analizar patrones de navegación o visualización.
- Utilizar técnicas de corrección de sesgo de exposición.

---

## Ausencia de demografía

### Observación

El dataset utilizado no incluye información demográfica detallada de los usuarios.

### Riesgo

No es posible establecer conclusiones sobre grupos poblacionales específicos ni evaluar comportamientos asociados a edad, género u otras características demográficas.

### Evidencia en el laboratorio

La recomendación se construye exclusivamente a partir de identificadores de usuario e historial de ratings.

### Posible mitigación

- Incorporar variables demográficas cuando estén disponibles.
- Evaluar el desempeño por segmentos de usuarios.
- Aplicar métricas de equidad y representatividad.

---

## Retroalimentación

### Observación

Las recomendaciones influyen en las decisiones futuras de los usuarios.

### Riesgo

Las películas recomendadas tienen una mayor probabilidad de ser consumidas y posteriormente valoradas, provocando que los datos futuros reflejen parcialmente las decisiones del propio sistema.

### Evidencia en el laboratorio

Las futuras valoraciones podrían utilizarse para reentrenar el modelo, reforzando patrones previamente aprendidos y amplificando determinados sesgos.

### Posible mitigación

- Mantener procesos periódicos de evaluación.
- Incorporar exploración controlada.
- Monitorear diversidad, cobertura y evolución del catálogo recomendado.

---

## Conclusión

Los sistemas recomendadores proporcionan recomendaciones útiles y personalizadas, pero también introducen riesgos relacionados con popularidad dominante, burbujas de filtro, sesgos de selección, ausencia de información demográfica y efectos de retroalimentación. Estos factores deben considerarse al interpretar los resultados obtenidos y al diseñar sistemas de recomendación para entornos reales.

````python
# ------------------------------------------------------------------------------
# Create a complete README for U03.E05
# ------------------------------------------------------------------------------

from pathlib import Path

target = Path(
    "/content/INF-8239-Ciencia-de-Datos-II/"
    "Unit_03_Generative_Algorithms/"
    "U03.E05/"
    "README.md"
)

readme = """
# U03.E05 - Recommender Systems

**Course:** INF-8239 Data Science II
**Unit:** 03 - Generative Algorithms and Features
**Student:** Francisco Pascual

---

# Project Overview

This project explores multiple recommendation strategies using the MovieLens dataset.

The work progresses from simple recommendation approaches to more advanced collaborative and hybrid recommender systems.

Implemented approaches:

1. Popularity-based recommendation
2. Content-based recommendation
3. Collaborative filtering with matrix factorization
4. Hybrid recommendation system

The project evaluates not only predictive accuracy but also ranking quality, catalog coverage, cold start behavior, and computational cost.

---

# Learning Objectives

The objective of this exercise is to understand:

- How recommendation systems operate.
- The difference between content-based and collaborative recommendation.
- Matrix factorization and latent factors.
- Offline evaluation procedures.
- Temporal train/test splitting.
- Ranking metrics.
- Cold start problems.
- Green AI tradeoffs between utility and computational cost.

---

# Dataset

## MovieLens Latest Small

MovieLens contains user interactions with movies through explicit ratings.

### Dataset Characteristics

| Metric | Value |
|----------|----------|
| Users | 610 |
| Movies | 9,742 |
| Ratings | 100,836 |
| Density | 0.01697 |

---

# LAB08 - Popularity and Content-Based Recommendation

## Popularity Baseline

The popularity baseline recommends items using:

- Average rating
- Rating count
- Bayesian smoothing

### Purpose

Provide a recommendation strategy even when user information is unavailable.

### Advantages

- Very simple.
- Easy to explain.
- Effective for cold-start situations.

### Limitations

- Not personalized.
- Popularity bias.

---

## Content-Based Recommendation

### Features Used

- Movie genres
- TF-IDF encoding
- Cosine similarity

### Workflow

Movie
→ Genres
→ TF-IDF
→ Cosine Similarity
→ Similar Movies

### Example

Toy Story

Generated recommendations:

- Toy Story 2
- Monsters Inc.
- Antz
- Turbo

### Conclusions

The model successfully identified similar movies based on genre information.

However:

- User preferences were ignored.
- Recommendations became highly specialized.
- No personalization was performed.

---

# LAB09 - Matrix Factorization and Hybrid Recommendation

## Collaborative Filtering

Collaborative filtering was implemented using matrix factorization.

The idea is to learn latent representations of:

- Users
- Movies

using historical ratings.

### Conceptual Workflow

User Ratings
→ Matrix Factorization
→ User Latent Vectors
→ Movie Latent Vectors
→ Rating Prediction
→ Recommendation Ranking

---

## Understanding Latent Factors

The algorithm learns a vector for each user and movie.

Example:

User 414

[0.82, 0.14, 0.73, ...]

Movie

[0.79, 0.11, 0.69, ...]

The interaction between both vectors is used to predict future ratings.

---

# Temporal Split

A temporal split was applied to avoid information leakage.

For each eligible user:

- Historical interactions were used for training.
- The most recent interaction was reserved for testing.

This simulates a realistic recommendation environment where only past data is available when making future recommendations.

---

# Evaluation Metrics

## RMSE

Measures rating prediction error.

Lower values indicate better approximation of observed ratings.

---

## HitRate@10

Measures whether the held-out item appears in the Top-10 recommendations.

Higher values indicate better ranking quality.

---

## Catalog Coverage

Measures how much of the movie catalog appears in recommendations.

Higher values indicate greater diversity.

---

# Hybrid Recommender

The hybrid system combines:

## Collaborative Component

Generated through matrix factorization.

## Content Component

Generated through:

- Genres
- TF-IDF
- Cosine similarity

### Hybrid Score

Hybrid Score =
α × Collaborative Score +
(1 − α) × Content Score

---

# Alpha Comparison

## Alpha = 0.25

Content-dominated configuration.

Results:

- RMSE: 1.0264
- HitRate@10: 0.0324
- Coverage: 0.1065

---

## Alpha = 0.75

Collaborative-dominated configuration.

Results:

- RMSE: 1.0264
- HitRate@10: 0.0375
- Coverage: 0.0775

---

## Selected Configuration

Alpha = 0.75

Reason:

The collaborative signal achieved the highest HitRate@10 while maintaining a similar computational cost.

---

# User Profile Evaluation

## Large History User

User ID: 414

Ratings:

2698

Characteristics:

- Strong collaborative signal.
- Stable latent representation.
- Highly personalized recommendations.

---

## Small History User

User ID: 406

Ratings:

20

Characteristics:

- Limited collaborative information.
- More dependence on content information.
- Less stable recommendations.

---

## New User

Characteristics:

- No ratings available.
- No collaborative representation can be built.

Fallback strategy:

- Rating average
- Rating count
- Popularity ranking

Recommendations are not personalized.

---

# Utility vs Cost Analysis

## Configuration A

- Factors: 10
- Epochs: 8
- RMSE: 1.0820
- HitRate@10: 0.0290
- Coverage: 0.0861
- Training Time: 9.25 s

## Configuration B

- Factors: 20
- Epochs: 12
- RMSE: 1.0264
- HitRate@10: 0.0375
- Coverage: 0.0775
- Training Time: 14.16 s

## Configuration C

- Factors: 40
- Epochs: 20
- RMSE: 1.0037
- HitRate@10: 0.0307
- Coverage: 0.1013
- Training Time: 22.45 s

---

# Pareto Analysis

Configuration A was dominated due to lower utility.

Configurations B and C remained on the Pareto frontier.

### Configuration B

Prioritizes:

- Better ranking quality.
- Lower computational cost.

### Configuration C

Prioritizes:

- Lower RMSE.
- Higher coverage.

---

# Green AI Perspective

The largest model was not automatically selected.

Although Configuration C achieved the lowest RMSE, it required:

- More factors
- More epochs
- More training time

while producing a lower HitRate@10 than Configuration B.

Therefore, Configuration B provided the best balance between utility and computational cost.

---

# Risks

## Popularity Bias

Popular items may dominate recommendations.

### Mitigation

Diversification strategies.

---

## Filter Bubble

Users may repeatedly receive similar content.

### Mitigation

Increase recommendation diversity.

---

## Selection Bias

Only observed interactions are available.

### Mitigation

Incorporate implicit feedback.

---

## Lack of Demographic Information

Population-level conclusions cannot be made.

### Mitigation

Include demographic attributes when available.

---

## Feedback Loops

Recommendations influence future training data.

### Mitigation

Continuous monitoring and periodic retraining.

---

# Final Conclusions

This project demonstrated the evolution of recommendation systems from popularity-based methods to hybrid recommendation architectures.

Key findings:

- Content-based recommendation identifies similar items.
- Collaborative filtering learns user preferences.
- Good rating prediction does not automatically imply good ranking quality.
- Hybrid recommenders balance multiple information sources.
- Larger models do not necessarily provide better practical performance.
- Green AI principles are useful when selecting recommendation models.

The final selected model was:

- 20 latent factors
- 12 training epochs
- Alpha = 0.75

because it achieved the best balance between recommendation quality and computational efficiency.
"""

target.write_text(readme, encoding="utf-8")

print("README updated successfully:")
print(target)
````

````python
# ==============================================================================
# AUTOMATIC COMMIT AND AUTHENTICATED PUSH
# Objective: Preserve empty directories, stage repository changes, create a
# commit when needed, push the current branch securely, and verify Git status.
# ==============================================================================

# Import os for environment and file-system operations.
import os

# Import stat to configure the temporary authentication script permissions.
import stat

# Import subprocess to execute Git commands and verify their results.
import subprocess

# Import Path for operating-system-independent path management.
from pathlib import Path

# Import userdata to retrieve the GitHub token from Colab Secrets.
from google.colab import userdata


# ------------------------------------------------------------------------------
# 1. DEFINE THE REPOSITORY CONFIGURATION
# ------------------------------------------------------------------------------

# Define the repository root directory.
REPO_PATH = Path(
    "/content/INF-8239-Ciencia-de-Datos-II"
)

# Define the Unit 02 directory.
UNIT_02_PATH = (
    REPO_PATH
    / "Unit_02_Natural_Language_Processing"
)

# Define the GitHub repository information.
GITHUB_USER = "pascual-francisco"
GITHUB_REPO = "INF-8239-Ciencia-de-Datos-II"

# Define the commit message.
COMMIT_MESSAGE = "chore: sync repository updates"

# Retrieve the GitHub token from Colab Secrets.
GITHUB_TOKEN = userdata.get("TOKEN")

# Define the clean GitHub remote URL.
CLEAN_REMOTE_URL = (
    f"https://github.com/{GITHUB_USER}/{GITHUB_REPO}.git"
)


# ------------------------------------------------------------------------------
# 2. VALIDATE THE REPOSITORY AND TOKEN
# ------------------------------------------------------------------------------

# Confirm that the repository directory exists.
if not REPO_PATH.exists():
    raise FileNotFoundError(
        f"The repository was not found:\n{REPO_PATH}"
    )

# Confirm that the repository contains a .git directory.
if not (REPO_PATH / ".git").exists():
    raise FileNotFoundError(
        f"The directory is not a Git repository:\n{REPO_PATH}"
    )

# Confirm that the GitHub token is available.
if not GITHUB_TOKEN:
    raise ValueError(
        "The GitHub token was not found. "
        "Create a Colab Secret named TOKEN and enable notebook access."
    )


# ------------------------------------------------------------------------------
# 3. DEFINE A REUSABLE GIT COMMAND FUNCTION
# ------------------------------------------------------------------------------

def run_git(
    arguments: list[str],
    *,
    check: bool = True,
    capture_output: bool = False,
    environment: dict[str, str] | None = None
) -> subprocess.CompletedProcess:
    """
    Execute a Git command from the repository root.

    Parameters
    ----------
    arguments:
        Git arguments without the initial 'git'.
    check:
        Raise an error when the command returns a nonzero exit code.
    capture_output:
        Capture stdout and stderr instead of displaying them directly.
    environment:
        Optional environment variables used by the Git command.
    """

    return subprocess.run(
        ["git", *arguments],
        cwd=REPO_PATH,
        check=check,
        text=True,
        capture_output=capture_output,
        env=environment
    )


# ------------------------------------------------------------------------------
# 4. PREPARE EMPTY DIRECTORIES
# ------------------------------------------------------------------------------

print("=" * 70)
print("PREPARING EMPTY DIRECTORIES")
print("=" * 70)

if UNIT_02_PATH.exists():

    created_gitkeep_files = []

    # Walk through every directory inside Unit 02.
    for directory_path, directory_names, file_names in os.walk(
        UNIT_02_PATH
    ):
        # A directory is empty when it has no folders and no files.
        if not directory_names and not file_names:
            gitkeep_path = Path(directory_path) / ".gitkeep"
            gitkeep_path.touch()
            created_gitkeep_files.append(gitkeep_path)

    if created_gitkeep_files:
        print(
            f"Created {len(created_gitkeep_files)} "
            ".gitkeep file(s)."
        )
    else:
        print("No empty directories required a .gitkeep file.")

else:
    raise FileNotFoundError(
        f"The Unit 02 directory was not found:\n{UNIT_02_PATH}"
    )


# ------------------------------------------------------------------------------
# 5. DISPLAY THE CURRENT CHANGES
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("CHANGES DETECTED")
print("=" * 70)

# Display modified, deleted, renamed, and untracked files.
run_git([
    "status",
    "--short",
    "--untracked-files=all"
])


# ------------------------------------------------------------------------------
# 6. STAGE THE REPOSITORY CHANGES
# ------------------------------------------------------------------------------

# Stage all changes that are not excluded by .gitignore.
run_git([
    "add",
    "-A"
])

print("\n" + "=" * 70)
print("FILES STAGED FOR COMMIT")
print("=" * 70)

# Retrieve the exact list of staged changes.
staged_result = run_git(
    [
        "diff",
        "--cached",
        "--name-status"
    ],
    capture_output=True
)

staged_output = staged_result.stdout.strip()

# Display the staged changes.
if staged_output:
    print(staged_output)
else:
    print("No files are currently staged.")


# ------------------------------------------------------------------------------
# 7. CREATE A COMMIT ONLY WHEN CHANGES EXIST
# ------------------------------------------------------------------------------

if staged_output:
    print("\nCreating commit...")

    run_git([
        "commit",
        "-m",
        COMMIT_MESSAGE
    ])

    print(f"Commit created: {COMMIT_MESSAGE}")

else:
    print("\nNo new changes to commit.")


# ------------------------------------------------------------------------------
# 8. IDENTIFY THE CURRENT BRANCH
# ------------------------------------------------------------------------------

branch_result = run_git(
    [
        "branch",
        "--show-current"
    ],
    capture_output=True
)

current_branch = branch_result.stdout.strip()

if not current_branch:
    raise RuntimeError(
        "Git could not determine the current branch."
    )


# ------------------------------------------------------------------------------
# 9. CONFIGURE A CLEAN REMOTE URL
# ------------------------------------------------------------------------------

# Keep the permanent origin URL free from credentials.
run_git([
    "remote",
    "set-url",
    "origin",
    CLEAN_REMOTE_URL
])


# ------------------------------------------------------------------------------
# 10. CREATE TEMPORARY NON-INTERACTIVE AUTHENTICATION
# ------------------------------------------------------------------------------

# Define a temporary Git authentication helper.
askpass_path = Path("/tmp/github_colab_askpass.sh")

# Create a temporary script that responds to Git credential prompts.
askpass_content = """#!/bin/sh
case "$1" in
    *Username*) printf '%s\\n' "$GITHUB_USERNAME" ;;
    *Password*) printf '%s\\n' "$GITHUB_TOKEN" ;;
    *) printf '\\n' ;;
esac
"""

# Save the authentication helper.
askpass_path.write_text(
    askpass_content,
    encoding="utf-8"
)

# Allow only the current user to execute or modify the helper.
askpass_path.chmod(
    stat.S_IRUSR
    | stat.S_IWUSR
    | stat.S_IXUSR
)

# Create an environment for the authenticated Git push.
push_environment = os.environ.copy()

# Tell Git which temporary authentication helper to use.
push_environment["GIT_ASKPASS"] = str(askpass_path)

# Prevent Git from opening an interactive credential prompt.
push_environment["GIT_TERMINAL_PROMPT"] = "0"

# Supply the GitHub username only through the process environment.
push_environment["GITHUB_USERNAME"] = GITHUB_USER

# Supply the GitHub token only through the process environment.
push_environment["GITHUB_TOKEN"] = GITHUB_TOKEN


# ------------------------------------------------------------------------------
# 11. PUSH THE CURRENT BRANCH
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("PUSHING TO GITHUB")
print("=" * 70)
print(f"Current branch: {current_branch}")

try:
    # Push the current branch and establish its upstream relationship.
    run_git(
        [
            "push",
            "--set-upstream",
            "origin",
            current_branch
        ],
        environment=push_environment
    )

finally:
    # Delete the temporary authentication helper after the push attempt.
    if askpass_path.exists():
        askpass_path.unlink()


# ------------------------------------------------------------------------------
# 12. DISPLAY THE LATEST COMMITS
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("LATEST COMMITS")
print("=" * 70)

# Display the five most recent commits.
run_git([
    "log",
    "--oneline",
    "--decorate",
    "-5"
])


# ------------------------------------------------------------------------------
# 13. VERIFY THE FINAL SYNCHRONIZATION STATUS
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SYNCHRONIZATION STATUS")
print("=" * 70)

# Display the final branch and working-tree status.
run_git([
    "status",
    "--short",
    "--branch"
])

print("\nRepository synchronized successfully.")
````
